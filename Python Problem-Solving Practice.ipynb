```python

# Logic:
# 1. Store the input string.
# 2. Initialize a counter to 0.
# 3. Check each character in the string.
# 4. Convert each character to lowercase to handle uppercase vowels.
# 5. If the character is a, e, i, o, or u, increase the counter by 1.
# 6. Print the total number of vowels.

string = "Education"
count = 0

for char in string:
    if char.lower() in "aeiou":
        count += 1

print(count)
```

```python

# Logic:
# 1. Store the input string.
# 2. Initialize a consonant counter to 0.
# 3. Check each character in the string.
# 4. Use isalpha() to check whether the character is a letter.
# 5. Use lower() to handle uppercase and lowercase letters.
# 6. If the character is not a vowel, increase the counter by 1.
# 7. Ignore spaces, digits, and special characters.
# 8. Print the total number of consonants.

string = "Python 123!"
count = 0

for char in string:
    if char.isalpha() and char.lower() not in "aeiou":
        count += 1

print(count)
```

```python

# Logic:
# 1. Store the input string.
# 2. Create an empty dictionary to store character frequencies.
# 3. Check each character and ignore spaces.
# 4. Increase the frequency of each character in the dictionary.
# 5. Find the character with the highest frequency.
# 6. Update the result only when a strictly higher frequency is found,
#    so the first character wins if there is a tie.
# 7. Print the most frequent character.

string = "banana"
frequency = {}

for char in string:
    if char != " ":
        frequency[char] = frequency.get(char, 0) + 1

most_frequent = ""
highest_count = 0

for char in string:
    if char != " " and frequency[char] > highest_count:
        highest_count = frequency[char]
        most_frequent = char

print(most_frequent)
```

```python



# Step 1: Store the input sentence
sentence = "Python is easy and Python is powerful"

# Step 2: Convert all words to lowercase
# This treats Python and python as the same word
sentence = sentence.lower()

# Step 3: Split the sentence into individual words
words = sentence.split()

# Step 4: Create an empty dictionary to store word counts
frequency = {}

# Step 5: Count the frequency of each word
for word in words:
    frequency[word] = frequency.get(word, 0) + 1

# Step 6: Initialize the most frequent word
most_frequent = words[0]
highest_count = 0

# Step 7: Find the word with the highest frequency
# Using > keeps the first word when frequencies are tied
for word in words:
    if frequency[word] > highest_count:
        highest_count = frequency[word]
        most_frequent = word

# Step 8: Print the result
print(most_frequent)
```

```python



# Step 1: Store the input sentence
sentence = "apple banana apple mango banana orange"

# Step 2: Convert the sentence to lowercase
# This treats Apple and apple as the same word
sentence = sentence.lower()

# Step 3: Split the sentence into individual words
words = sentence.split()

# Step 4: Create a variable to store the unique word
first_unique = None

# Step 5: Check each word in its original order
for word in words:
    # Check whether the word appears exactly once
    if words.count(word) == 1:
        # Store the first word that appears only once
        first_unique = word

        # Stop searching after finding the first unique word
        break

# Step 6: Print the result
if first_unique is not None:
    print(first_unique)
else:
    print("No unique word found")
```

```python



# Step 1: Store the input sentence
sentence = "Python makes coding useful Python"

# Step 2: Convert the sentence to lowercase
# This treats Python and python as the same word
sentence = sentence.lower()

# Step 3: Split the sentence into individual words
words = sentence.split()

# Step 4: Initialize the longest unique word
longest_word = ""

# Step 5: Check each word in its original order
for word in words:
    # Step 6: Check whether the word appears exactly once
    if words.count(word) == 1:

        # Step 7: Update only if the word is longer
        # Using > keeps the first word when lengths are tied
        if len(word) > len(longest_word):
            longest_word = word

# Step 8: Print the result
if longest_word:
    print(longest_word)
else:
    print("No unique word found")
```
